### Task 1 - Index creation

>Set up a vector storage system. It can be any vector DBMS you want, or a SQL DBMS with vector support (for example, PostgreSQL with the pgvector add-on)

PostreSQL with the pgvector add-on was selected as a vector storage system. To initialize it via Docker, execute the `docker-compose up -d` command in the terminal within the directory containing the docker-compose.yml file. 

>Set up an index management API. The API must contain the following endpoints:
>- `/add_document` - accepts a document, splits it into chunks,vectorizes the chunks, and adds them to the vector storage;
>- `/query` - accepts a query vector and returns the top Kclosest chunks.

`api_task1.py` is a simple FastAPI application implementing the required endpoints. To start the server, execute the following command in the terminal: `uvicorn api_task1:app --port 8000 --reload`

>Gather any document library you want. For example: scientific papers, lab reports from previous courses, etc. Populate the vector storage with the gathered documents.

```python
import os
import requests
from pypdf import PdfReader

folder_path = "library"
api_url = "http://127.0.0.1:8000/add_document"

for filename in os.listdir(folder_path):
    if filename.endswith(".pdf"):
        file_path = os.path.join(folder_path, filename)
        
        # Read the PDF text
        reader = PdfReader(file_path)
        extracted_text = ""
        for page in reader.pages:
            extracted_text += page.extract_text() + " "
            
        if extracted_text.strip():
            # Send it to your Task 1 API
            requests.post(api_url, json={"text": extracted_text})
            print(f"Successfully processed and embedded: {filename}")
        else:
            print(f"Skipped empty or unreadable file: {filename}")
```

### Task 2 - Q&A Assistant

Gemini was selected for its free benefits for students.

>Create an LLM prompt template that contains:
>- a task definition;
>- placeholders for context chunks;
>- a placeholder for the user query.

>Set up the Q&A Assistant API. The API must contain:
>- a `/query` endpoint, which accepts the user's question and returns the answer based on the retrieved documents.


`api_task2.py` is a simple FastAPI application with the required endpoint. To setup server, open a terminal and run a command `uvicorn api_task2:app --port 8001 --reload`

Testing endpoint requests:

```python
user_question = "Побудова мережі Петрі"

response = requests.post("http://127.0.0.1:8001/query", json={"question": user_question})
data = response.json()

print("Question:", user_question)
print("\nAnswer:", data["answer"])
print("\nContext used by Gemini:", data["context_used"])
```
